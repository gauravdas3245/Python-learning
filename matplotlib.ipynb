```python
import matplotlib.pyplot as plt
category=['Apple','Banana','Mango','Jerry']
value=[20,30,10,40]
plt.bar(category,value,color='red', edgecolor='black')
plt.xlabel('Fruits Type')
plt.ylabel('Fruits Quantity')
plt.title('Bar Plt of Session Fruits')
plt.show()
```

```python
category=['Apple', 'Banana', 'Mango', 'Jerry']
value=[20,30,10,40]
plt.pie(value, labels=category)
plt.title('Session Fruits')
plt.show()
```

```python
catagory=['Apple', 'Banana', 'Mango', 'Jerry']
value=[20,30,10,40]
myexplode=[0.1,0,0,0]

plt.pie(value, labels=catagory, explode=myexplode, shadow=True)
plt.title('Session Fruits')
plt.show()
```

```python
data = [10, 12, 15, 18, 20, 22, 25, 30, 35, 40, 50]
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
plt.boxplot(data)

plt.title("Box Plot and IQR")
plt.ylabel("Values")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
t=np.arange(0.1,5.0,0.2)
plt.plot(t,t**2,'b--',label='^2')
plt.plot(t,t**2.2,'rs',label='^2.2')
plt.plot(t,t**2.5,'g^',label='^2.5')
plt.plot(t,t**2.8,'yd',label='^2.8')
plt.xlabel('numbers')
plt.ylabel('squares')
plt.title('Myplot')
plt.grid()
plt.legend()
plt.show()
```
