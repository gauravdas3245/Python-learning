# EDA

```python
plt.figure(figsize=(10,5))
sns.heatmap(df.isnull(),cbar=False)
plt.title("Missing values")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.heatmap(df.corr(),annot=True)
plt.title("Missing values")
plt.show()
```

```python
plt.figure(figsize=(10,7))
sns.countplot(data=df,x="LoanApproved")
plt.title("LoanApproved")
plt.show()
```

```python
plt.figure(figsize=(10,7))
sns.countplot(data=df,x="EmploymentStatus")
plt.title("EmploymentStatus")
plt.show()
```

```python
df.head()
```

```python
numerical_columns=[
    "MonthlyIncome",
    "Age",
    "CreditScore",
    "LoanAmount",
    "LoanTermMonths",
    "Dependents"
]

df[numerical_columns].hist(figsize=(10,7),bins=20,color='red')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df[numerical_columns])
plt.title("Box Plot of Numerical Columns")
plt.show()
```

```python
Q1 = df["MonthlyIncome"].quantile(0.25)
Q3 = df["MonthlyIncome"].quantile(0.75)
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df, x="LoanApproved", y="MonthlyIncome")
plt.title("Monthly Income vs Loan Approved")
plt.xlabel("Loan Approved")
plt.ylabel("Monthly Income")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df, x="LoanApproved", y="LoanAmount")
plt.title("Loan Amount vs Loan Approved")
plt.xlabel("Loan Approved")
plt.ylabel("Loan Amount")
plt.show()
```

```python
sns.pairplot(df[numerical_columns + ["LoanApproved"]], hue="LoanApproved")
plt.show()
```
